## 1. What is Python?

Python is a programming language: we write instructions and Python runs them. It is widely used for automation, data analysis, machine learning and AI applications. The `print()` function displays a result. Text inside quotes is called a string.

```python
print("Hello, AI learners!")
print("Let us learn Python.")
```

## 2. Variables and types

A variable is a name for a value. `=` assigns a value. Names are case-sensitive; use descriptive names such as `student_name`. A value has a **type**, which affects what operations make sense. `type()` shows its type.

```python
student_name = "Asha"
age = 25
score = 88.5
is_active = True
print(student_name, age, score, is_active)
print(type(student_name), type(age), type(score), type(is_active))
```

## 3. Common built-in data types

| Type | Example | Meaning |
|---|---|---|
| `str` | `"hello"` | Text |
| `int` | `42` | Whole number |
| `float` | `3.5` | Decimal number |
| `bool` | `True` | True or False |
| `NoneType` | `None` | No value is currently provided |
| `list` | `[1, 2]` | Ordered, changeable collection |
| `tuple` | `(1, 2)` | Ordered collection that cannot be changed in place |
| `set` | `{1, 2}` | Collection of unique values |
| `dict` | `{"name": "Asha"}` | Key–value pairs |
| `range` | `range(3)` | Sequence of integers, often used in loops |
| `complex` | `2 + 3j` | Complex number; useful in some scientific work |
| `bytes` | `b"hi"` | Immutable binary data |
| `bytearray` | `bytearray(b"hi")` | Changeable binary data |
| `memoryview` | `memoryview(b"hi")` | A view of binary data |

Focus on the first nine types for now. We will use binary and complex types later only if a task needs them. Use `set()` for an empty set; `{}` is an empty dictionary.

```python
values = ["hello", 42, 3.5, True, None, [1, 2], (1, 2), {1, 2}, {"name": "Asha"}, range(3), 2 + 3j, b"hi", bytearray(b"hi"), memoryview(b"hi")]
for value in values:
    print(repr(value), "->", type(value).__name__)
```

### More examples: types in everyday work

A number can be stored as text when it comes from a form. The value `None` can mean information is missing. A Boolean is useful when an answer has two states. Here `type(...).__name__` shows a short type name.

```python
# This age is text because it is inside quotes.
age_from_form = "25"
# Convert the numeric text to a whole number before adding.
age_next_year = int(age_from_form) + 1
# None represents a value that has not been provided.
phone_number = None
# A Boolean can record whether a customer has paid.
has_paid = False
print("Next year:", age_next_year)
print("Phone provided:", phone_number is not None)
print("Payment status:", has_paid, type(has_paid).__name__)
```

### More collections

A dictionary uses a key to find a value. A tuple keeps its items in order; it cannot be changed in place. A set stores unique items. We sort a set only for a predictable display.

```python
# Named fields make a record easy to read.
student = {"name": "Ravi", "score": 75}
print(student["name"], student["score"])
# A tuple is a fixed grouping, such as width and height.
image_size = (640, 480)
print("Width:", image_size[0])
# Repeated values appear only once in a set.
tags = {"delivery", "refund", "delivery"}
print("Unique tags:", sorted(tags))
```

### Numbers, operators and conversion

`+`, `-`, `*`, `/` perform arithmetic. `//` gives floor division, `%` gives a remainder, and `**` gives a power. `int()` and `float()` can convert valid numeric text. Invalid input such as `int("hello")` raises an error.

```python
price = 120
quantity = 3
print("Total:", price * quantity)
print("Average:", 10 / 3, "Floor:", 10 // 3, "Remainder:", 10 % 3)
print("Power:", 2 ** 3)
print(int("25") + 5)
print(float("3.5") * 2)
```

### Collections in a small customer record

A list can change. A tuple is useful for a fixed pair. A set removes duplicates; its order should not be relied on. A dictionary stores values under named keys.

```python
topics = ["Python", "AI"]
size = (640, 480)
tags = {"delivery", "refund", "delivery"}
customer = {"name": "Asha", "orders": 2}
print(topics, size, sorted(tags))
print(customer["name"], customer.get("city", "Not provided"))
```

## 4. Strings: working with text

Strings are sequences of characters. Indexing starts at 0; `-1` means the last character. A slice like `[0:4]` includes index 0 and stops before index 4. Strings are immutable: methods return a new string; they do not change the original.

```python
message = "  Hello Python  "
clean = message.strip()
print(repr(message), repr(clean))
print(clean[0], clean[-1], clean[0:5])
print("Length:", len(clean))
```

Common methods: `strip()` removes surrounding whitespace; `lower()` and `upper()` change case; `replace()` returns replaced text; `split()` splits text into a list; `join()` joins strings. `in` checks whether text appears in another string.

```python
text = "  Delivery, Refund, Support  "
clean = text.strip()
print(clean.lower())
print(clean.upper())
print(clean.replace("Refund", "Returns"))
parts = clean.split(", ")
print(parts)
print(" | ".join(parts))
print("Delivery" in clean)
name = "Asha"
print(f"Hello, {name}!")
```

### More string examples

`find()` returns the first position of matching text, or `-1` when it is missing. `count()` counts non-overlapping matches. `startswith()` and `endswith()` return Booleans. `replace()` does not change the original text.

```python
# Start with a support message.
message = "Order delayed; order is being checked"
# lower() returns a new string for a case-insensitive check.
lower_message = message.lower()
print("Position of order:", lower_message.find("order"))
print("Number of mentions:", lower_message.count("order"))
print("Starts with Order:", message.startswith("Order"))
print("Ends with checked:", message.endswith("checked"))
# The original remains unchanged when we assign the replacement.
updated = message.replace("delayed", "shipped")
print("Updated:", updated)
print("Original:", message)
```

```python
# Split a simple comma-separated line into pieces.
line = " Python, ML, GenAI "
parts = line.split(",")
clean_parts = []
for part in parts:
    # strip() removes extra spaces from the ends of each piece.
    clean_parts.append(part.strip())
print(clean_parts)
# join() puts a chosen separator between the items.
print(" | ".join(clean_parts))
```

## 5. Lists: working with groups

A list is ordered and changeable. Use `append` to add one item, `extend` to add several, `insert` to add at a position, `remove` to remove the first matching value, and `pop` to remove and return an item. `sort` changes a list; `sorted` returns a new sorted list.

```python
messages = ["hello", "help", "thanks"]
print(messages[0], messages[-1], messages[0:2])
messages.append("bye")
messages.extend(["order", "refund"])
messages.insert(1, "welcome")
messages.remove("help")
removed = messages.pop()
print("Removed:", removed)
print("Messages:", messages)
print("Count:", len(messages))
```

```python
scores = [80, 50, 90, 50]
print("Number of 50s:", scores.count(50))
print("Index of 90:", scores.index(90))
print("Sorted copy:", sorted(scores))
print("Original:", scores)
scores.sort()
print("After sort:", scores)
```

### More list examples: changing and copying

`append()` changes the original list. `copy()` creates a separate, shallow list. For a flat list of strings, this means changes to the copied list do not change the original. `in` checks if a value is present.

```python
topics = ["Python", "ML", "LLMs"]
# Make a separate list before adding another topic.
weekend_topics = topics.copy()
weekend_topics.append("Agents")
print("Original:", topics)
print("Copy:", weekend_topics)
print("Contains ML:", "ML" in topics)
# A slice starts at index 1 and stops before index 3.
print("Topics at indices 1 and 2:", topics[1:3])
```

```python
# Store one score per student.
scores = [55, 80, 35, 90]
total = 0
for score in scores:
    # Add the current score to the running total.
    total += score
print("Total:", total)
# Check for an empty list before dividing by its length.
if scores:
    print("Average:", total / len(scores))
```

## 6. Decisions: if, elif and else

Comparisons such as `==`, `!=`, `<`, `>=` produce `True` or `False`. `=` assigns; `==` compares. Indentation groups the code inside a condition. `and` needs both conditions to be true, `or` needs at least one.

```python
score = 72
if score >= 80:
    print("Excellent")
elif score >= 50:
    print("Pass")
else:
    print("Practise and try again")
print("Eligible:", score >= 50 and score <= 100)
```

## 7. For loops and range

A `for` loop processes each item in a sequence. `range(stop)` starts at zero and stops **before** `stop`. `range(start, stop, step)` lets you choose the start and step. A negative step counts down.

```python
for topic in ["Python", "ML", "LLMs"]:
    print("Learning:", topic)

for number in range(1, 6):
    print(number)
print("Even numbers:", list(range(0, 10, 2)))
print("Countdown:", list(range(3, 0, -1)))
```

### More `for` and `range` examples

Use `range(len(items))` when you need the index, although looping over items is usually simpler. `enumerate` gives both the position and item. Nested loops repeat the inner loop for every outer item.

```python
topics = ["Python", "AI", "Agents"]
# enumerate supplies a position and a value; start=1 is for display.
for position, topic in enumerate(topics, start=1):
    print(position, topic)
# Each day receives each task.
for day in ["Saturday", "Sunday"]:
    for task in ["Learn", "Practise"]:
        print(day, "-", task)
```

```python
# range has an excluded stop value.
for number in range(2, 11, 2):
    print("Even number:", number)
# Counting backwards needs a negative step.
print("Countdown:", list(range(5, 0, -1)))
```

## 8. While loops, break and continue

A `while` loop repeats while its condition is true. Update the counter so the loop can stop. `continue` skips the current iteration; `break` ends the loop.

```python
attempt = 1
while attempt <= 3:
    print("Attempt", attempt)
    attempt += 1

for message in ["hello", " ", "help", "stop", "later"]:
    if not message.strip():
        continue
    if message == "stop":
        break
    print("Process:", message)
```

### More `while` examples

A `while` loop is helpful when the number of repetitions depends on a condition. Track what changes on each pass and make sure the condition can eventually become false.

```python
# Count completed messages until the target is reached.
processed = 0
target = 3
while processed < target:
    processed += 1  # Update the counter so the loop can finish.
    print("Processed:", processed)
print("All done")
```

```python
# Add numbers until their total reaches at least 10.
number = 1
total = 0
while total < 10:
    total += number       # Add the current number.
    print("Added", number, "total is", total)
    number += 1           # Move on to the next number.
print("Final total:", total)
```

## 9. Real-life example: prepare support messages

Before passing messages to a later AI application, we can remove surrounding spaces and skip empty messages. This is basic data preparation; no model is called here.

```python
raw_messages = ["  Where is my order? ", " ", "I need a refund"]
clean_messages = []
for message in raw_messages:
    clean = message.strip()
    if clean:
        clean_messages.append(clean)
print("Ready:", clean_messages)
print("Count:", len(clean_messages))
```

- `TypeError`: check the value types; `"5"` is text and `5` is a number.
- `IndexError`: check that the list has an item at that index.
- `ValueError`: the value may not be suitable for a conversion or operation.
